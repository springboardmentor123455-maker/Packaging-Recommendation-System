```python
#Imported important libraries

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
#Loaded the dataset

df  = pd.read_csv('data.csv')
df.head
```

```python
#Check dataset structure Verify columns & data types

df.info()
```

```python
#Select ML features

feature_cols = [
    "Strength_MPa",
    "Weight_Capacity_kg",
    "Biodegradability_Score",
    "Recyclability_Percentage",
    "CO2_Emission_Score_kg"
]

X = df[feature_cols]
X.head()
```

```python
#Create target variable for CO2 Emission

y_cost = df["Price_INR"]
y_cost.head()
```

```python
#Generate target CO₂ Impact Prediction

y_co2 = df["CO2_Emission_Score_kg"]
y_co2.head()
```

```python
# Split the dataset into training and testing sets for cost prediction

X_train_cost, X_test_cost, y_train_cost, y_test_cost = train_test_split(X, y_cost, test_size=0.2, random_state=42)
print(X_train_cost.shape, X_test_cost.shape)
```

```python
# Split the dataset into training and testing sets for CO2 Emission Prediction

X_train_co2, X_test_co2, y_train_co2, y_test_co2 = train_test_split(X, y_co2, test_size=0.2, random_state=42)
print(X_train_co2.shape, X_test_co2.shape)
```

```python
#Scaling (Data Pipeline step) Normalize data for ML models

scaler = StandardScaler()

X_train_cost_scaled = scaler.fit_transform(X_train_cost)
X_test_cost_scaled = scaler.transform(X_test_cost)

X_train_co2_scaled = scaler.fit_transform(X_train_co2)
X_test_co2_scaled = scaler.transform(X_test_co2)
```

```python
#verify the shapes of the scaled data

print("Cost model training data shape:", X_train_cost_scaled.shape)
print("CO2 model training data shape:", X_train_co2_scaled.shape)
```
