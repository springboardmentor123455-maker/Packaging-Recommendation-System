```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
df = df.fillna("Unknown")
```

```python
print(df.columns)
```

```python
# Normalized numerical features

from sklearn.preprocessing import MinMaxScaler

numeric_cols = [
    "Strength_MPa",
    "Weight_Capacity_kg",
    "Biodegradability_Score",
    "CO2_Emission_Score_kg",
    "Recyclability_Percentage",
    "Price_INR"
]

scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("Normalized numerical features:")
df[numeric_cols].describe()
```

```python
df[numeric_cols].hist(figsize=(12,8), bins=20)
plt.suptitle("Distribution of Normalized Numerical Features")
plt.show()
```

```python
for col in numeric_cols:
    plt.figure(figsize=(6,3))
    plt.hist(df[col], bins=20)
    plt.title(f"Normalized Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

```

```python
# Encoding categorical features

df = pd.get_dummies(
    df,
    columns=[
        "Material_Type",
        "Product_Category",
        "Availability_Status"
    ],
    drop_first=True
)

print("Data after encoding categorical features:")
df.head()
```

```python
category_counts = df.filter(like="Product_Category_").sum()

plt.figure(figsize=(8,4))
category_counts.plot(kind="bar")
plt.ylabel("Count")
plt.title("Encoded Product Category Distribution")
plt.show()
```

```python
# CO2 Impact Index

df["CO2_Impact_Index"] = 1 / (df["CO2_Emission_Score_kg"] + 1)

df[["CO2_Emission_Score_kg", "CO2_Impact_Index"]].head()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(
    df["CO2_Emission_Score_kg"],
    df["CO2_Impact_Index"]
)
plt.xlabel("CO₂ Emission Score")
plt.ylabel("CO₂ Impact Index")
plt.title("CO₂ Emission vs CO₂ Impact Index")
plt.show()
```

```python
# Cost Efficiency Index

df["Cost_Efficiency_Index"] = df["Recyclability_Percentage"] / (df["Price_INR"] + 1)

df[["Recyclability_Percentage", "Price_INR", "Cost_Efficiency_Index"]].head()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(
    df["Price_INR"],
    df["Cost_Efficiency_Index"]
)
plt.xlabel("Price (Normalized)")
plt.ylabel("Cost Efficiency Index")
plt.title("Cost vs Cost Efficiency Index")
plt.show()
```

```python
# Material Suitability Score

df["Material_Suitability_Score"] = (
    0.4 * df["Strength_MPa"] +
    0.3 * df["Biodegradability_Score"] +
    0.3 * df["Recyclability_Percentage"]
)

df[["Strength_MPa", "Biodegradability_Score",
    "Recyclability_Percentage", "Material_Suitability_Score"]].head()
```

```python
plt.figure(figsize=(7,4))
plt.hist(df["Material_Suitability_Score"], bins=30)
plt.xlabel("Material Suitability Score")
plt.ylabel("Frequency")
plt.title("Distribution of Material Suitability Score")
plt.show()
```
