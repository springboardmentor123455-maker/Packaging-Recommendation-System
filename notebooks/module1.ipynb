```python
import mysql.connector
import pandas as pd
```

```python
connection = mysql.connector.connect(
    host="127.0.0.1",
    user="root",
    port="3306",
    password="omkapse@45",
    database="eco_packaging"
)

cursor = connection.cursor()
print("Connected to MySQL successfully!")
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS materials (
    Material_ID VARCHAR(50),
    Material_Type VARCHAR(100),
    Product_Category VARCHAR(100),
    Strength_MPa DOUBLE,
    Weight_Capacity_kg DOUBLE,
    Biodegradability_Score DOUBLE,
    CO2_Emission_kg DOUBLE,
    Recyclability_Percent DOUBLE,
    Source VARCHAR(100),
    Certification VARCHAR(100),
    Cost_INR DOUBLE,
    Notes VARCHAR(255)
);
"""

cursor.execute(create_table_query)
connection.commit()
print("Table created successfully!")
```

```python
df = pd.read_csv("data.csv")
df.head()
```

```python
df = df.where(pd.notnull(df), None)
```

```python
insert_query = """
INSERT INTO materials VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)
"""

for _, row in df.iterrows():
    cursor.execute(insert_query, tuple(row))

connection.commit()
print("Data inserted into MySQL successfully!")
```

```python
cursor.execute("SELECT COUNT(*) FROM materials")
count = cursor.fetchone()[0]
print("Total records in table:", count)
```

```python
query = """
SELECT Material_Type, Biodegradability_Score, Recyclability_Percent, Cost_INR
FROM materials
ORDER BY Biodegradability_Score DESC, Recyclability_Percent DESC
LIMIT 10;
"""
pd.read_sql(query, connection)
```

```python
query = """
SELECT Material_Type, Biodegradability_Score, Recyclability_Percent, Cost_INR
FROM materials
WHERE Product_Category = 'Textiles'
ORDER BY Biodegradability_Score DESC;
"""
pd.read_sql(query, connection)
```

```python
query = """
SELECT Material_Type, Cost_INR
FROM materials
ORDER BY Cost_INR ASC
LIMIT 10;
"""
pd.read_sql(query, connection)
```
