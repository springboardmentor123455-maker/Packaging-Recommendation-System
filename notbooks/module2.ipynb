```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.fillna(df.mean(numeric_only=True))
```

```python
df = df.fillna("Unknown")
```

```python
print(df.columns)
```

```python
# Normalized numerical features

from sklearn.preprocessing import MinMaxScaler

numeric_cols = [
    "Strength_MPa",
    "Weight_Capacity_kg",
    "Biodegradability_Score",
    "CO2_Emission_Score_kg",
    "Recyclability_Percentage",
    "Price_INR"
]

scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("Normalized numerical features:")
df[numeric_cols].describe()
```

```python
# Encoding categorical features

df = pd.get_dummies(
    df,
    columns=[
        "Material_Type",
        "Product_Category",
        "Availability_Status"
    ],
    drop_first=True
)

print("Data after encoding categorical features:")
df.head()
```

```python
# CO2 Impact Index

df["CO2_Impact_Index"] = 1 / (df["CO2_Emission_Score_kg"] + 1)

df[["CO2_Emission_Score_kg", "CO2_Impact_Index"]].head()
```

```python
# Cost Efficiency Index

df["Cost_Efficiency_Index"] = df["Recyclability_Percentage"] / (df["Price_INR"] + 1)

df[["Recyclability_Percentage", "Price_INR", "Cost_Efficiency_Index"]].head()
```

```python
# Material Suitability Score

df["Material_Suitability_Score"] = (
    0.4 * df["Strength_MPa"] +
    0.3 * df["Biodegradability_Score"] +
    0.3 * df["Recyclability_Percentage"]
)

df[["Strength_MPa", "Biodegradability_Score",
    "Recyclability_Percentage", "Material_Suitability_Score"]].head()
```
